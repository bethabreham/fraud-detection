```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df_fraud = pd.read_csv('../data/raw/Fraud_Data.csv')
print(f"Fraud_Data shape: {df_fraud.shape}")
df_fraud.head()

# Check data types
print(df_fraud.dtypes)

# Check missing values
print(df_fraud.isnull().sum())

# Check class imbalance
print(df_fraud['class'].value_counts())
print(f"Fraud percentage: {df_fraud['class'].mean()*100:.4f}%")

# Plot class distribution
plt.figure(figsize=(6, 4))
df_fraud['class'].value_counts().plot(kind='bar')
plt.title('Class Distribution (E-commerce)')
plt.xlabel('Class (0=Legit, 1=Fraud)')
plt.ylabel('Count')
plt.show()

# Amount distribution
plt.figure(figsize=(10, 6))
plt.hist(df_fraud['purchase_value'], bins=50, edgecolor='black')
plt.title('Purchase Value Distribution')
plt.xlabel('Purchase Value ($)')
plt.ylabel('Frequency')
plt.show()
```
