```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df_credit = pd.read_csv('../data/raw/creditcard.csv')
print(f"creditcard shape: {df_credit.shape}")
df_credit.head()

# Check missing values
print(df_credit.isnull().sum())

# Check class imbalance
print(df_credit['Class'].value_counts())
print(f"Fraud percentage: {df_credit['Class'].mean()*100:.4f}%")

# Plot class distribution
plt.figure(figsize=(6, 4))
df_credit['Class'].value_counts().plot(kind='bar')
plt.title('Class Distribution (Credit Card)')
plt.xlabel('Class (0=Legit, 1=Fraud)')
plt.ylabel('Count')
plt.show()

# Amount distribution by class
plt.figure(figsize=(10, 6))
sns.boxplot(x='Class', y='Amount', data=df_credit)
plt.title('Transaction Amount by Class')
plt.yscale('log')
plt.show()
```
